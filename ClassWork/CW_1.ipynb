```python
from pycoingecko import CoinGeckoAPI
import pandas as pd  

cg = CoinGeckoAPI()

btc_data = cg.get_coin_market_chart_by_id(id='bitcoin', vs_currency='usd', days=10)

btc_data
```

```python
prices = btc_data['prices']
df = pd.DataFrame(prices, columns=['Timestamp', 'Price'])
df['Timestamp'] = pd.to_datetime(df['Timestamp'], unit='ms')
print(df.head())
```

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

# URL & Parameters
url = "https://api.coingecko.com/api/v3/coins/markets?vs_currency=usd&order=market_cap_desc&per_page=10&page=1&sparkline=false&price_change_percentage=7d"
params = {
    "vs_currency": "usd",
    "order": "market_cap_desc",
    "per_page": 10,
    "page": 1,
    "sparkline": "false",
    "price_change_percentage": "7d"
}

# GET Request
response = requests.get(url, params=params)
data = response.json()

# แปลงเป็น DataFrame
df = pd.DataFrame(data)

# เลือกเฉพาะคอลัมน์ที่ต้องการ
df_filtered = df[['id', 'symbol', 'name', 'market_cap_rank', 'market_cap', 'price_change_percentage_7d_in_currency']]

# แสดง DataFrame
print(df_filtered)

# วาดกราฟ Market Cap
plt.figure(figsize=(10, 6))
plt.bar(df_filtered['name'], df_filtered['market_cap'])
plt.xticks(rotation=45)
plt.title('Top 10 Cryptos by Market Cap')
plt.ylabel('Market Cap (USD)')
plt.tight_layout()
plt.show()

# วาดกราฟ % เปลี่ยนแปลงใน 7 วัน
plt.figure(figsize=(10, 6))
plt.bar(df_filtered['name'], df_filtered['price_change_percentage_7d_in_currency'])
plt.xticks(rotation=45)
plt.title('7d Price Change (%) of Top 10 Cryptos')
plt.ylabel('Price Change (%)')
plt.axhline(0, color='gray', linestyle='--')
plt.tight_layout()
plt.show()
```
